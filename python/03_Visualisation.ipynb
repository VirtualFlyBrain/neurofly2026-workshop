## Route A — `vfb_connect`
VFB returns neurons already in shared template space, so co-visualisation is direct. Load skeletons/meshes and plot with `navis`. For a no-code 3D view, VFB's website [Circuit Browser](https://virtualflybrain.org/docs/website-features/circuitbrowser/) and the MCP/chat image responses work in-browser.

````python
import navis
# Download aligned skeletons (SWC) for a few IDs, then plot with navis.
ids = ['VFB_jrchjtdb', 'VFB_fw035286']  # two DA1 lPNs: hemibrain + FlyWire (from module 01)
vfb.get_images(ids, template='JRC2018Unisex', image_folder='./swc', image_type='swc')
neurons = navis.read_swc('./swc')
navis.plot3d(neurons)
# For a no-code 3D view, the same neurons open in the VFB website via vfb.get_vfb_link(ids, template=...).
````
