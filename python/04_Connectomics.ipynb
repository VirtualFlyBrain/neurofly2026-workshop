## Route A — `vfb_connect`
Query weighted synaptic partners and trace short pathways. VFB wraps connectivity across datasets; neuPrint (module reference) gives dataset-native queries.

```python
# Connectivity between two neuron TYPES, filtered by minimum synaptic weight (verified API).
# Here: the classic olfactory->memory step, DA1 lPN -> Kenyon cell (mushroom body).
conn = vfb.get_connected_neurons_by_type(
    upstream_type='adult antennal lobe projection neuron DA1 lPN',
    downstream_type='Kenyon cell',
    weight=10)
conn[['upstream_neuron_name', 'weight', 'downstream_neuron_name', 'downstream_class', 'up_data_source']].head(10)
# Returns real connections with per-synapse weights across datasets (e.g. BANC, hemibrain).
# Set only upstream_type (or only downstream_type) to widen the net; also
# vfb.get_neurons_upstream_of(...) / get_neurons_downstream_of(...) for type-level partners.
```
