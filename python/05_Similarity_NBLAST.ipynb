## Route A — `vfb_connect`
NBLAST ([Costa et al. 2016](https://doi.org/10.1016/j.neuron.2016.06.005)) scores morphological similarity. VFB precomputes cross-dataset NBLAST, so you can find the closest cell in another connectome — now including BANC, male-CNS and optic-lobe.

````python
# Morphologically similar neurons to a query, across datasets (verified current API).
hits = vfb.get_similar_neurons('VFB_jrchjtdb', similarity_score='NBLAST_score')  # DA1_lPN_R (hemibrain)
# The chosen metric is the input; the returned score column is simply 'score'.
hits.sort_values('score', ascending=False).head(10)
# Top hits are other DA1 lPNs: in-dataset first (hemibrain, ~0.75-0.80), then cross-dataset
# matches in FlyWire and FAFB (~0.66-0.68) — the same cell type recovered by morphology alone.
# Related: vfb.get_potential_drivers(neuron) finds GAL4/split lines by morphology.
````
