## Route A — `vfb_connect`
Resolve two individuals, compare their ontology types, and confirm with morphology (NBLAST, module 05). Shared type + high NBLAST = strong cross-dataset identity. `vfb.terms()` fetches TermInfo; `vfb.get_vfb_link()` builds a website link to eyeball them side by side.

```python
# Two real DA1 lPN individuals from different connectomes:
#   VFB_jrchjtdb  DA1_lPN_R, hemibrain (bodyId 1734350908)
#   VFB_fw035286  AL.MB_CA.83, FlyWire v783
# Fetch both in ONE call — vfb.terms() returns a VFBTerms collection, and .summary
# gives a tidy DataFrame. (Calling terms() per-id and concatenating works but is
# slower and emits a "VFBTerm only has one item" warning for every field.)
pair = vfb.terms(['VFB_jrchjtdb', 'VFB_fw035286'])

# Inspect the parent (type) labels to see whether they share a classification.
pair.summary[['ID', 'Name', 'Parents', 'Datasets']]
```
