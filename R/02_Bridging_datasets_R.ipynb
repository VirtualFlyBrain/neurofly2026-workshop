# 02 · Bridging datasets (R)

**Problem P2: is this the same cell in another dataset?**

R companion to the [workshop page](https://workshop.virtualflybrain.org/sessions/session-2-bridging-identity/?route=r) — same `vfb_connect` engine as the Python track, driven through `reticulate`. All outputs below were verified live in Aug 2026.

## Setup — once per session

Colab: make sure the runtime is **R** (Runtime → Change runtime type → R). The install below takes a couple of minutes; the first `import` connects to VFB services (~2 min) and is fast afterwards.

```r
# Once per session
install.packages("reticulate", quiet = TRUE)   # needs a current reticulate (pandas-3 support)
reticulate::py_install("vfb-connect", pip = TRUE)
```

```r
library(reticulate)
vfb <- import("vfb_connect")$vfb     # first call connects, ~2 min
cat("Connected.\n")
```

```r
m <- vfb$get_similar_neurons("VFB_jrchjtdb",
                             similarity_score = "NBLAST_score",
                             return_dataframe = TRUE)
fw <- m[grepl("flywire", m$source_id), ]
head(fw[, c("id", "score", "label")], 3)
```

**Verified output:** best FlyWire match `VFB_fw036329` (AL.MB_CA.111) at **0.68**, annotated with the same cell types as the query — morphology score + shared ontology typing is the bridge. In-dataset hemibrain sisters score 0.75–0.80, so expect a cross-dataset penalty.

natverse users: `vfbconnectr::read.neurons.vfb()` pulls the matches into `nat` for your own NBLAST runs.
