# 05 · Similarity / NBLAST (R)

**Problem P5: morphologically similar neurons across datasets.**

R companion to the [workshop page](https://workshop.virtualflybrain.org/problems/p5-similarity-/-nblast/?route=r) — same `vfb_connect` engine as the Python track, driven through `reticulate`. All outputs below were verified live in Aug 2026.

## Setup — once per session

Colab: make sure the runtime is **R** (Runtime → Change runtime type → R). The install below takes a couple of minutes; the first `import` connects to VFB services (~2 min) and is fast afterwards.

```r
# Once per session
install.packages("reticulate", quiet = TRUE)   # needs a current reticulate (pandas-3 support)
reticulate::py_install("vfb-connect", pip = TRUE)
```

```r
library(reticulate)
vfb <- import("vfb_connect")$vfb     # first call connects, ~2 min
cat("Connected.\n")
```

```r
m <- vfb$get_similar_neurons("VFB_jrchjtdb",
                             similarity_score = "NBLAST_score",
                             return_dataframe = TRUE)
nrow(m)
head(m[, c("id", "score", "label")], 7)
```

**Verified output:** 107 neighbours — hemibrain sisters 0.80–0.75, then FAFB and FlyWire tied at **0.68**. Scores fall smoothly with no clean threshold; a morphologically similar *different* type (DL3 lPN) sits at ≈0.65. Sort, inspect, judge.
