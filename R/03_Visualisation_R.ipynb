# 03 · Visualisation (R)

**Problem P3: see neurons together in a common template space.**

R companion to the [workshop page](https://workshop.virtualflybrain.org/sessions/session-3-visualisation/?route=r) — same `vfb_connect` engine as the Python track, driven through `reticulate`. All outputs below were verified live in Aug 2026.

## Setup — once per session

Colab: make sure the runtime is **R** (Runtime → Change runtime type → R). The install below takes a couple of minutes; the first `import` connects to VFB services (~2 min) and is fast afterwards.

```r
# Once per session
install.packages("reticulate", quiet = TRUE)   # needs a current reticulate (pandas-3 support)
reticulate::py_install("vfb-connect", pip = TRUE)
```

```r
library(reticulate)
vfb <- import("vfb_connect")$vfb     # first call connects, ~2 min
cat("Connected.\n")
```

```r
da1 <- vfb$term("VFB_jrchjtdb")
da1$load_skeleton(template = "JRC2018Unisex")
da1$skeleton                        # navis TreeNeuron, 4847 nodes
```

```r
# Interactive 3D over the standard template
da1$plot3d(template = "JRC2018Unisex", include_template = TRUE)
```

```r
# Overlay its strongest partners in the same space
partners <- da1$downstream_partners(weight = 30L)
da1$plot_partners(partners[1:3], template = "JRC2018Unisex")
```

Always pass the template explicitly — that is what keeps every neuron in the same coordinate space. For a fully native R workflow, `vfbconnectr::read.neurons.vfb("VFB_jrchjtdb", template = "JRC2018Unisex")` returns a `nat::neuronlist` for `nat`/`rgl` plotting.
