# 07 · Putting it together (R)

**Problem P7: the whole chain — region → type → instance → partners → matches → expression.**

R companion to the [workshop page](https://workshop.virtualflybrain.org/sessions/session-7-putting-it-together/?route=r) — same `vfb_connect` engine as the Python track, driven through `reticulate`. All outputs below were verified live in Aug 2026.

## Setup — once per session

Colab: make sure the runtime is **R** (Runtime → Change runtime type → R). The install below takes a couple of minutes; the first `import` connects to VFB services (~2 min) and is fast afterwards.

```r
# Once per session
install.packages("reticulate", quiet = TRUE)   # needs a current reticulate (pandas-3 support)
reticulate::py_install("vfb-connect", pip = TRUE)
```

```r
library(reticulate)
vfb <- import("vfb_connect")$vfb     # first call connects, ~2 min
cat("Connected.\n")
```

```r
kc_types <- vfb$term("Kenyon cell")$children               # 37 subtypes
inst <- vfb$get_instances("alpha/beta Kenyon cell",
                          return_dataframe = TRUE)          # 7212 records
hb  <- inst[grepl("hb", sapply(inst$data_source, paste, collapse=",")), ]
kc  <- vfb$term(hb$id[[1]])                                 # e.g. KCab-m_R
kc$name
```

```r
kc$load_skeleton(template = "JRC2018Unisex")
kc$plot3d(template = "JRC2018Unisex", include_template = TRUE)
```

```r
for (p in kc$downstream_partners(weight = 20L)) print(p)
```

**Verified payoff** — the textbook mushroom-body cast: APL_R (34), DPM_R (31), MBON06 (23).

```r
m <- vfb$get_similar_neurons(kc$id, similarity_score = "NBLAST_score",
                             return_dataframe = TRUE)
e <- vfb$get_transcriptomic_profile("alpha/beta Kenyon cell",
                                    return_dataframe = TRUE)
c(nblast_neighbours = nrow(m), expression_rows = nrow(e))
```
