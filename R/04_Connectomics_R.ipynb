# 04 · Connectomics (R)

**Problem P4: strongest partners and pathways.**

R companion to the [workshop page](https://workshop.virtualflybrain.org/problems/p4-connectomics/?route=r) — same `vfb_connect` engine as the Python track, driven through `reticulate`. All outputs below were verified live in Aug 2026.

## Setup — once per session

Colab: make sure the runtime is **R** (Runtime → Change runtime type → R). The install below takes a couple of minutes; the first `import` connects to VFB services (~2 min) and is fast afterwards.

```r
# Once per session
install.packages("reticulate", quiet = TRUE)   # needs a current reticulate (pandas-3 support)
reticulate::py_install("vfb-connect", pip = TRUE)
```

```r
library(reticulate)
vfb <- import("vfb_connect")$vfb     # first call connects, ~2 min
cat("Connected.\n")
```

```r
da1 <- vfb$term("VFB_jrchjtdb")
for (p in da1$downstream_partners(weight = 20L)[1:5]) print(p)
```

**Verified output:** v2LN30_R (73), lLN2T_c (61), DA1_vPN_R (61), lLN2P_b (47), lLN2T_c #2 (43). Note the `L` suffix — reticulate passes R integers as Python ints only with `20L`.

```r
kc <- vfb$get_connected_neurons_by_type(upstream_type = "DA1 lPN",
                                        downstream_type = "Kenyon cell",
                                        weight = 10L, return_dataframe = TRUE)
nrow(kc)          # expect 1132
```

`get_connected_neurons_by_type` defaults to `exclude_dbs = ['hb', 'fafb']` (avoids double-counting the overlapping volumes), so these 1,132 pairs are FlyWire + MaleCNS + BANC.
