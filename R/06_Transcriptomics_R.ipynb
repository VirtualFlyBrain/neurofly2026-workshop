# 06 · Transcriptomics (R)

**Problem P6: expression profile of a cell type.**

R companion to the [workshop page](https://workshop.virtualflybrain.org/problems/p6-transcriptomics/?route=r) — same `vfb_connect` engine as the Python track, driven through `reticulate`. All outputs below were verified live in Aug 2026.

## Setup — once per session

Colab: make sure the runtime is **R** (Runtime → Change runtime type → R). The install below takes a couple of minutes; the first `import` connects to VFB services (~2 min) and is fast afterwards.

```r
# Once per session
install.packages("reticulate", quiet = TRUE)   # needs a current reticulate (pandas-3 support)
reticulate::py_install("vfb-connect", pip = TRUE)
```

```r
library(reticulate)
vfb <- import("vfb_connect")$vfb     # first call connects, ~2 min
cat("Connected.\n")
```

```r
e <- vfb$get_transcriptomic_profile("alpha/beta Kenyon cell",
                                    return_dataframe = TRUE)   # ~40 s
nrow(e)
head(e[, c("gene", "level", "extent")], 5)
```

**Verified output:** 18,234 gene × cluster rows with expression `level`, `extent` (fraction of cells expressing) and a real publication per row (Lu 2023, Davie 2018, …), ready for `dplyr`/`ggplot2`.

Profile a *specific subtype* — the bare parent class "Kenyon cell" expands 37 subtypes and takes 15–20 minutes. Check a type has data first via the `hasScRNAseq` search facet.
