# 01 · Discovery (R)

**Problem P1: find every instance of a neuron type across all datasets.**

R companion to the [workshop page](https://workshop.virtualflybrain.org/sessions/session-1-discovery/?route=r) — same `vfb_connect` engine as the Python track, driven through `reticulate`. All outputs below were verified live in Aug 2026.

## Setup — once per session

Colab: make sure the runtime is **R** (Runtime → Change runtime type → R). The install below takes a couple of minutes; the first `import` connects to VFB services (~2 min) and is fast afterwards.

```r
# Once per session
install.packages("reticulate", quiet = TRUE)   # needs a current reticulate (pandas-3 support)
reticulate::py_install("vfb-connect", pip = TRUE)
```

```r
library(reticulate)
vfb <- import("vfb_connect")$vfb     # first call connects, ~2 min
cat("Connected.\n")
```

```r
df <- vfb$get_instances("DA1 lPN", return_dataframe = TRUE)
nrow(df)
table(sapply(df$data_source, paste, collapse = ","))
head(df[, c("id", "label")], 5)
```

**Verified output:** 68 records — BANC 18, FlyWire 15, FAFB 15, MaleCNS 13, hemibrain 7. These are registered *images*, not cells (~8 DA1 lPNs per hemisphere biologically). Columns holding lists (like `data_source`) arrive as R list-columns; `sapply(..., paste)` flattens them.
